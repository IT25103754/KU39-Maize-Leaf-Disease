## 3. Preprocessing & EDA
### Technique and justification
Repeated copies can leak identical image content across train and test. We compare file SHA-256 and orientation-corrected RGB pixel hashes. Pixel hashes include dimensions. Conflicting labels are quarantined; consistent exact duplicates retain the first sorted path. This does not detect all near duplicates.

### Implementation and actual outputs

```python
# Member technique: identify exact pixel duplicates and conflicting labels.
valid = manifest[manifest.valid == True].copy()
label_counts = valid.groupby('rgb_sha256').label.nunique()
conflict_hashes = set(label_counts[label_counts > 1].index)
consistent = valid[~valid.rgb_sha256.isin(conflict_hashes)].sort_values('path')
member_clean = consistent.drop_duplicates('rgb_sha256', keep='first')
summary = pd.DataFrame({
    'check': ['Raw file duplicates beyond first', 'Conflicting-label images',
              'Consistent pixel duplicates beyond first', 'Retained images'],
    'count': [int(valid.duplicated('file_sha256').sum()),
              int(valid.rgb_sha256.isin(conflict_hashes).sum()),
              int(consistent.duplicated('rgb_sha256').sum()), len(member_clean)]})
display(summary)
assert set(member_clean.path) == set(cleaned.path)
summary.to_csv(ROOT/'results/outputs/member2_duplicate_summary.csv', index=False)
```

```python
# The shared implementation uses StratifiedGroupKFold(seed=39, n_splits=5).
# Fold 0 -> test; fold 1 -> validation; folds 2,3,4 -> training.
# This is a NEW split for preparation, NOT recovered original training evidence.
from sklearn.model_selection import StratifiedGroupKFold
work = member_clean.reset_index(drop=True).copy()
work['fold'] = -1
folds = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=SEED)
for fold, (_, held) in enumerate(folds.split(work, work.label, groups=work.rgb_sha256)):
    work.loc[held,'fold'] = fold
work['split'] = work.fold.map({0:'test', 1:'validation', 2:'train', 3:'train', 4:'train'})
groups = {s:set(work.loc[work.split == s,'rgb_sha256']) for s in ['train','validation','test']}
for a,b in [('train','test'), ('train','validation'), ('validation','test')]:
    overlap = len(groups[a] & groups[b])
    print(a, '/', b, 'exact pixel group overlap:', overlap)
    assert overlap == 0
assert work.set_index('path')['split'].sort_index().equals(cleaned.set_index('path')['preparation_split'].sort_index())
display(pd.crosstab(work.label, work.split).reindex(CLASSES))
```
