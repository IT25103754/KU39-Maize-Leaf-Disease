## 3. Preprocessing & EDA
### Technique and justification
Image files can be unreadable or stored under unexpected labels. We decode each file, validate its folder label, record the error and preserve the raw archive. Missing pixel data is not handled by inventing replacement leaf images.

### Implementation and actual outputs

```python
# Member technique: independently check every image for successful decoding.
import io
from collections import Counter
checks = []
with zipfile.ZipFile(ROOT/'data/raw/archive.zip') as archive:
    for name in sorted(archive.namelist()):
        if Path(name).suffix.lower() not in {'.jpg', '.jpeg', '.png'}: continue
        try:
            if Path(name).parent.name not in CLASSES:
                raise ValueError('Unknown folder label')
            with Image.open(io.BytesIO(archive.read(name))) as im:
                im.load()  # actually decode pixels, not only the header
                checks.append({'path': name, 'status': 'valid', 'reason': ''})
        except Exception as exc:
            checks.append({'path': name, 'status': 'excluded', 'reason': str(exc)})
integrity = pd.DataFrame(checks)
integrity.to_csv(ROOT/'results/outputs/member1_integrity_check.csv', index=False)
display(integrity.status.value_counts().rename('images').to_frame())
assert integrity.status.eq('valid').sum() == manifest.valid.sum()
print('Independent decode check matches the shared manifest.')
display(manifest.groupby(['image_format','mode']).size().rename('images').to_frame())
```
