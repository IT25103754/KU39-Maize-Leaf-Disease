## 3. Preprocessing & EDA
### Technique and justification
Training augmentation introduces controlled visual variations without adding new independent leaves. We demonstrate horizontal flipping, a small rotation and modest brightness adjustment using only the new training subset. Validation/test images receive deterministic preprocessing only.

### Implementation and actual outputs

```python
# Member technique: reproducible augmentation of a training example only.
from PIL import ImageEnhance
train = cleaned[cleaned.preparation_split == 'train']
source = train.sort_values('path').iloc[0]
image = load_image(ROOT, source.path).resize((224,224),Image.Resampling.BILINEAR)
def member_augment(image, seed):
    rng=np.random.default_rng(seed)
    flipped=image.transpose(Image.Transpose.FLIP_LEFT_RIGHT)
    angle=float(rng.uniform(-12,12))
    rotated=flipped.rotate(angle,resample=Image.Resampling.BILINEAR)
    return ImageEnhance.Brightness(rotated).enhance(float(rng.uniform(.85,1.15)))
a=np.asarray(member_augment(image,SEED))
b=np.asarray(member_augment(image,SEED))
np.testing.assert_array_equal(a,b)
assert source.preparation_split == 'train'
assert a.shape == (224,224,3)
print('Selected source split:',source.preparation_split)
print('Reproducibility check: identical seed gives identical pixels.')
print('Mean absolute pixel change:',float(np.abs(a.astype(float)-np.asarray(image).astype(float)).mean()))
print('No synthetic variants were added to validation or test.')
```
